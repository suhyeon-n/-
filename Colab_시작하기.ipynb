```
from collections import deque

vertices = ['s', "v'", 'v1', 'v2', 'v3', 'v4', 't']
edges = [
    ('s', 'v1', 16), ('s', 'v2', 13), ('v1', "v'", 10), ('v1', 'v3', 12),
    ("v'", 'v2', 10), ('v2', 'v1', 4), ('v2', 'v4', 14), ('v3', 'v2', 9),
    ('v3', 't', 20), ('v4', 'v3', 7), ('v4', 't', 4)
]

adj = {v: [] for v in vertices}
cap = {}
for u, v, c in edges:
    adj[u].append(v)
    adj[v].append(u)
    cap[(u, v)] = cap.get((u, v), 0) + c
    cap[(v, u)] = cap.get((v, u), 0)

def bfs(src, snk, parent):
    visited = {src}
    q = deque([src])
    while q:
        u = q.popleft()
        for v in adj[u]:
            if v not in visited and cap[(u, v)] > 0:
                visited.add(v)
                parent[v] = u
                if v == snk:
                    return True
                q.append(v)
    return False

def edmonds_karp(src, snk):
    parent = {}
    max_flow = 0
    while bfs(src, snk, parent):
        flow = float('inf')
        v = snk
        while v != src:
            flow = min(flow, cap[(parent[v], v)])
            v = parent[v]
        max_flow += flow
        v = snk
        while v != src:
            u = parent[v]
            cap[(u, v)] -= flow
            cap[(v, u)] += flow
            v = u

    S = {src}
    q = deque([src])
    while q:
        u = q.popleft()
        for v in adj[u]:
            if v not in S and cap[(u, v)] > 0:
                S.add(v)
                q.append(v)

    s_list = [v for v in vertices if v in S]
    t_list = [v for v in vertices if v not in S]
    return max_flow, s_list, t_list

max_flow, S, T = edmonds_karp('s', 't')
print(f"[에드몬드-카프 알고리즘 결과] Maximum flow : {max_flow}")
print(f"[에드몬드-카프 알고리즘 결과] Minimum cut : S = {{{', '.join(S)}}}, T = {{{', '.join(T)}}}")
```
