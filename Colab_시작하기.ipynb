```
W = 14
items = [
    ('A', 2, 12),
    ('B', 1, 10),
    ('C', 3, 20),
    ('D', 2, 15),
    ('E', 4, 24),
    ('F', 5, 28),
    ('G', 6, 32),
    ('H', 3, 19)
]

n = len(items)
dp = [[0] * (W + 1) for _ in range(n + 1)]

for i in range(1, n + 1):
    name, w, v = items[i - 1]
    for j in range(W + 1):
        if w <= j:
            dp[i][j] = max(dp[i - 1][j], dp[i - 1][j - w] + v)
        else:
            dp[i][j] = dp[i - 1][j]

max_val = dp[n][W]

res = []
cw = W
for i in range(n, 0, -1):
    if dp[i][cw] != dp[i - 1][cw]:
        name, w, v = items[i - 1]
        res.append(name)
        cw -= w
res.reverse()

print("1. 9x15 DP 테이블:")
for r in dp:
    print(r)

print(f"\n2. 배낭의 최대 허용 무게를 초과하지 않는 최대 가치:\n{max_val}")

print(f"\n3. 최대 가치를 얻기 위해 선택된 물건들의 목록 및 총 무게:")
print(f"목록: {res}")
print(f"총 무게: {W - cw}") # 또는 선택된 물건들의 무게 합
```
