```
# 과제 #1: 힙 정렬(Heap sort) 및 기수 정렬(Radix sort) 구현

# 입력 데이터
arr = [88, 4, 485, 299, 98, 175, 58, 935, 258, 593, 42, 26, 386, 683, 582, 147, 743, 130, 704, 1]


# -------------------------------------------------------------
# 1. 힙 생성 (Build Max-Heap) 함수 및 실행
# -------------------------------------------------------------
def heapify(arr, n, i):
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left

    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

def build_heap(arr):
    heap_arr = arr.copy()
    n = len(heap_arr)
    # 마지막 자식이 있는 부모 노드부터 루트 노드까지 heapify 수행
    for i in range(n // 2 - 1, -1, -1):
        heapify(heap_arr, n, i)  # <--- 세 번째 인자를 n으로 수정!
    return heap_arr

# 1번 출력: 주어진 배열로부터 생성된 힙
heap_array = build_heap(arr)
print("1. 1x20 입력 배열로부터 생성된 1x20 힙의 배열:")
print(heap_array)
print()


# -------------------------------------------------------------
# 2. 힙 정렬 (Heap Sort) 함수 및 실행
# -------------------------------------------------------------
def heap_sort(arr):
    # 힙 생성
    sorted_arr = build_heap(arr)
    n = len(sorted_arr)

    # 힙에서 요소를 하나씩 꺼내어 정렬
    for i in range(n - 1, 0, -1):
        sorted_arr[0], sorted_arr[i] = sorted_arr[i], sorted_arr[0] # 루트와 마지막 노드 교환
        heapify(sorted_arr, i, 0) # 줄어든 힙에 대해 heapify 재수행

    return sorted_arr

# 2번 출력: 힙 정렬로 정렬된 배열
heap_sorted_array = heap_sort(arr)
print("2. 힙 정렬로 정렬된 1x20 배열:")
print(heap_sorted_array)
print()


# -------------------------------------------------------------
# 3. 기수 정렬 (Radix Sort) 함수 및 실행
# -------------------------------------------------------------
def counting_sort_for_radix(arr, exp):
    n = len(arr)
    output = [0] * n
    count = [0] * 10

    # 자릿수(exp) 기준 카운트 계산
    for i in range(n):
        index = arr[i] // exp
        count[index % 10] += 1

    # 누적 합 계산
    for i in range(1, 10):
        count[i] += count[i - 1]

    # 안정 정렬(Stable Sort)을 위해 뒤에서부터 출력 배열 구성
    for i in range(n - 1, -1, -1):
        index = arr[i] // exp
        output[count[index % 10] - 1] = arr[i]
        count[index % 10] -= 1

    return output

def radix_sort(arr):
    radix_arr = arr.copy()
    max_val = max(radix_arr)

    # 1의 자리부터 시작하여 가장 큰 수의 자릿수까지 자릿수를 높여가며 정렬
    exp = 1
    while max_val // exp > 0:
        radix_arr = counting_sort_for_radix(radix_arr, exp)
        exp *= 10

    return radix_arr

# 3번 출력: 기수 정렬로 정렬된 배열
radix_sorted_array = radix_sort(arr)
print("3. 기수 정렬로 정렬된 1x20 배열:")
print(radix_sorted_array)
```

```
W = 14
items = [
    ('A', 2, 12),
    ('B', 1, 10),
    ('C', 3, 20),
    ('D', 2, 15),
    ('E', 4, 24),
    ('F', 5, 28),
    ('G', 6, 32),
    ('H', 3, 19)
]

n = len(items)
dp = [[0] * (W + 1) for _ in range(n + 1)]

for i in range(1, n + 1):
    name, w, v = items[i - 1]
    for j in range(W + 1):
        if w <= j:
            dp[i][j] = max(dp[i - 1][j], dp[i - 1][j - w] + v)
        else:
            dp[i][j] = dp[i - 1][j]

max_val = dp[n][W]

res = []
cw = W
for i in range(n, 0, -1):
    if dp[i][cw] != dp[i - 1][cw]:
        name, w, v = items[i - 1]
        res.append(name)
        cw -= w
res.reverse()

print("1. 9x15 DP 테이블:")
for r in dp:
    print(r)

print(f"\n2. 배낭의 최대 허용 무게를 초과하지 않는 최대 가치:\n{max_val}")

print(f"\n3. 최대 가치를 얻기 위해 선택된 물건들의 목록 및 총 무게:")
print(f"목록: {res}")
print(f"총 무게: {W - cw}") # 또는 선택된 물건들의 무게 합
```
