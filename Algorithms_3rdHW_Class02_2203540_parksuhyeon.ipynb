```
graph = {
    'a': ['b', 'd'],
    'b': ['c'],
    'c': [],
    'd': ['e', 'f'],
    'e': ['f', 'g'],
    'f': ['g', 'h'],
    'g': [],
    'h': []
}

for node in graph:
    graph[node].sort()

def topological_sort_dfs(graph):
    visited = set()
    post_order = []

    def dfs(node):
        visited.add(node)
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                dfs(neighbor)
        post_order.append(node)

    for node in graph:
        if node not in visited:
            dfs(node)

    return post_order[::-1]

topo_result = topological_sort_dfs(graph)

print("[위상 정렬 결과 - DFS 기반]")
print(" -> ".join(topo_result))
print("리스트 형태:", topo_result)
```
